```python
# Old column names
old_cols = ['Country', 'Year', 'Status', 'Life expectancy ', 'Adult Mortality',
       'infant deaths', 'Alcohol', 'percentage expenditure', 'Hepatitis B',
       'Measles ', ' BMI ', 'under-five deaths ', 'Polio', 'Total expenditure',
       'Diphtheria ', ' HIV/AIDS', 'GDP', 'Population',
       ' thinness  1-19 years', ' thinness 5-9 years',
       'Income composition of resources', 'Schooling']

# New clean column names
new_cols = [
    'country', 'year', 'status', 'life_expectancy', 'adult_mortality',
    'infant_deaths', 'alcohol', 'percentage_expenditure', 'hepatitis_b',
    'measles', 'bmi', 'under_five_deaths', 'polio', 'total_expenditure',
    'diphtheria', 'hiv_aids', 'gdp', 'population',
    'thinness_10_19_years', 'thinness_5_9_years',
    'income_composition_of_resources', 'schooling'
]

# Rename using dictionary
df = df.rename(columns=dict(zip(old_cols, new_cols)))

# check result
print(df.columns)
```
